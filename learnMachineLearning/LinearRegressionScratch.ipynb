```python
import numpy as np 


X = np.random.randint(1, 100, size = (20000, 5))
Y = np.random.randint(1,100, size = (20000,1))


#b = np.random.rand(1,1)
w = np.random.rand(5,1)

#ensuring X Y , w had the same cols, so it can works
assert X.shape[0] == Y.shape[0]
assert X.shape[1] == w.shape[0]

#Start Multiplying it 
N = np.dot(X, w)
squaredDifferetiaion = np.power(Y - N, 2)
mse = np.mean(squaredDifferetiaion)

#Computing derivatives of loss func respected to N
dldn = -2 * (Y - N)
dldn
#Computing derivatives of N respected to weight
dndw = X.T
dndw
#Computing derivatives of loss function respected to weights
dldw = np.dot(dndw, dldn)
```
